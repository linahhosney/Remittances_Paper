```r
# Filter to SPLITA 2 for remittances qs
df_2 <- df %>%
  filter(splita == 2)

print(nrow(df_2))
print(ncol(df_2))
```

```r
# Check country variable type immediately after SPLITA filter
cat("Country class:", class(df_2$country), "\n")
table(df_2$country)
```

```r
df_2 <- df_2 %>%
  select(
    # Identifiers
    country,
    splita,

    # Independent variable
    remit = Q1017,

    # Dependent variables — non-electoral
    protest = Q502_2,
    meeting_petition = Q502_1,
    civic_member = Q501,
    volunteer = Q501A,
    charity = Q266,

    # Dependent variables — electoral
    # vote_national = Q301A,
    vote_local = Q301C,
    rally = Q302,

    # Key interaction variable
    corrupt = Q210,

    # Control variables
    education = Q1003,
    income = Q1016,
    econ_perception = Q101,
    age = Q1001,
    gender = Q1002,
    employed = Q1005,
    migration_intent = Q104,
    pol_interest = Q404
  )

# Verify
dim(df_2)
names(df_2)
head(df_2, 5)
```

```r
# Check raw values for each variable
vars_to_check <- c("Q1017", "Q502_2", "Q502_1", "Q501", "Q501A",
                   "Q266", "Q301C", "Q302", "Q210", "Q1003",
                   "Q1016", "Q101", "Q1001", "Q1002", "Q1005",
                   "Q104", "Q404")

# Run on original df filtered to splita 2 only
df_splita2 <- df %>% filter(splita == 2)

for(var in vars_to_check) {
  cat("\n===", var, "===\n")
  print(table(df_splita2[[var]], useNA = "always"))
}
```

```r
# Check missing value codes across all variables
cat("=== BEFORE CLEANING: Count of 98, 99, 99999 ===\n")
df_2 %>%
  summarise(across(everything(),
                   ~sum(. %in% c(98, 99, 99999),
                   na.rm = TRUE))) %>%
  pivot_longer(everything(),
               names_to = "variable",
               values_to = "count") %>%
  filter(count > 0) %>%
  arrange(desc(count)) %>%
  print()

# Replace 98, 99, and 99999 with NA
df_2 <- df_2 %>%
  mutate(across(everything(),
                ~ifelse(. %in% c(98, 99, 99999),
                        NA, .)))

# Verify they are gone
cat("\n=== AFTER CLEANING: Count of 98, 99, 99999 (should all be 0) ===\n")
df_2 %>%
  summarise(across(everything(),
                   ~sum(. %in% c(98, 99, 99999),
                   na.rm = TRUE))) %>%
  pivot_longer(everything(),
               names_to = "variable",
               values_to = "count") %>%
  filter(count > 0) %>%
  print()

cat("All clean if nothing printed above\n")
```

```r
#recoding binary variables
df_2 <- df_2 %>%
  mutate(
    # Independent variable
    remit = case_when(
      remit %in% c(1, 2, 3) ~ 1,
      remit == 4 ~ 0,
      TRUE ~ NA_real_
    ),

    # Non-electoral DVs
    protest = case_when(
      protest %in% c(1, 2) ~ 1,
      protest == 3 ~ 0,
      TRUE ~ NA_real_
    ),

    meeting_petition = case_when(
      meeting_petition %in% c(1, 2) ~ 1,
      meeting_petition == 3 ~ 0,
      TRUE ~ NA_real_
    ),

    civic_member = case_when(
      civic_member == 1 ~ 1,
      civic_member == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    volunteer = case_when(
      volunteer == 1 ~ 1,
      volunteer == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    # Electoral DVs
    vote_local = case_when(
      vote_local == 1 ~ 1,
      vote_local == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    rally = case_when(
      rally == 1 ~ 1,
      rally == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    # Gender: 0=male, 1=female
    gender = case_when(
      gender == 1 ~ 0,
      gender == 2 ~ 1,
      TRUE ~ NA_real_
    ),

    # Migration intention: 1=yes, 0=no
    migration_intent = case_when(
      migration_intent == 1 ~ 1,
      migration_intent == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    # Employment: 1=employed, 0=not employed
    employed = case_when(
      employed %in% c(1, 2, 3) ~ 1,
      employed %in% c(4, 5, 6, 7) ~ 0,
      TRUE ~ NA_real_
    ),

    # Charity: 1=Yes, 0=No
    charity = case_when(
      charity == 1 ~ 1,
      charity == 2 ~ 0,
      TRUE ~ NA_real_
    ),

    # Corruption binary: 1=high (large/medium), 0=low (small/not at all)
    # Original: 1=large extent, 2=medium, 3=small, 4=not at all
    corrupt = case_when(
      corrupt %in% c(1, 2) ~ 1,  # high corruption perception
      corrupt %in% c(3, 4) ~ 0,  # low corruption perception
      TRUE ~ NA_real_
    )
  )

# Verify key variables
table(df_2$remit, useNA = "always")
table(df_2$protest, useNA = "always")
table(df_2$gender, useNA = "always")
table(df_2$migration_intent, useNA = "always")
table(df_2$employed, useNA = "always")
table(df_2$corrupt, useNA = "always")
```

```r
# Create remit_by_country data
remit_by_country <- df_clean %>%
  mutate(country_name = case_when(
    country == "5"  ~ "Egypt",
    country == "7"  ~ "Iraq",
    country == "8"  ~ "Jordan",
    country == "10" ~ "Lebanon",
    country == "11" ~ "Libya",
    country == "13" ~ "Morocco",
    country == "15" ~ "Palestine",
    country == "19" ~ "Sudan",
    country == "21" ~ "Tunisia",
    country == "22" ~ "Yemen",
    TRUE ~ as.character(country)
  )) %>%
  group_by(country_name) %>%
  summarise(
    n = n(),
    pct_remit = round(mean(remit == 1) * 100, 1),
    .groups = "drop"
  ) %>%
  arrange(desc(pct_remit))

# Verify
print(remit_by_country)
```

```r
ggplot(remit_by_country, aes(x = reorder(country_name, pct_remit),
                              y = pct_remit)) +
  geom_point(size = 4, color = "black") +
  geom_segment(aes(x = reorder(country_name, pct_remit),
                   xend = reorder(country_name, pct_remit),
                   y = 0, yend = pct_remit),
               color = "gray60", linewidth = 0.5) +
  geom_text(aes(label = paste0(pct_remit, "%")),
            hjust = -0.4, size = 3.5) +
  coord_flip() +
  labs(x = "", y = "") +
  theme_minimal() +
  theme(
    panel.grid = element_blank(),
    axis.text = element_text(size = 11, color = "black"),
    axis.ticks = element_blank()
  ) +
  ylim(0, max(remit_by_country$pct_remit) + 6)

ggsave("remittances_by_country.png", width = 7, height = 5, dpi = 300)
cat("Figure saved\n")
```
